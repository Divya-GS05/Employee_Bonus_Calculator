```python
Emp_Name = input("Enter the employee name: ")
Salary = int(input("Enter the salary of the employee: "))
Performance = input("Enter the performance of the employee: (Excellent, Good, Average, Poor)")

def calculate_bonus(Salary, Performance):
  if Performance == "Excellent":
    return Salary * 0.2
  elif Performance == "Good":
    return Salary * 0.1
  elif Performance == "Average":
    return Salary * 0.05
  elif Performance == "Poor":
    return 0
  else:
    print("Invalid Rating")
    return 0
Bonus = calculate_bonus(Salary, Performance)

def employee_report():
  print("\nEmployee Name: ", Emp_Name)
  print("Monthly Salary: ",Salary)
  print("Performance Rating: ",Performance)
  print("Monthly Bonus: ",Bonus)
  print("Annual Salary with Bonus: ", Salary * 12 + Bonus)
employee_report()


```
